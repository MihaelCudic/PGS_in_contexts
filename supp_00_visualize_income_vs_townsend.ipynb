```python
"""
    Code to visualize the relationship between deprivation and income
    Supplemental Figure S1
"""

import pandas as pd
import numpy as np
import random
import time

from options.options import Options
import util.util as util
import util.pre_process as pre
import util.bootstrap_tools as btool

import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as plticker
import numpy as np

import torch
from scipy.stats import norm
```

```python
opt = Options()
opt.initialize()

covars_df = pd.read_csv(opt.env_file, sep='\t', index_col='IID')
```

```python
import seaborn as sns

x_tickers = ['≤£18k', '£18k - £31k', '£31k - £52k', '£52k - £100k', '≥£100k']

# Create violin plot
plt.figure(figsize=(8, 6))
sns.violinplot(x='Household_Income', y='Townsend_Deprivation_Index', data=covars_df)
plt.xlabel('Household Income Level', fontsize=14)
plt.ylabel('Townsend Deprivation Index', fontsize=14)
plt.title('UKB: Deprivation vs. Income', fontsize=16)
plt.xticks(ticks=range(5), rotation=30, labels=x_tickers, fontsize=12)
plt.yticks(fontsize=12)
plt.savefig('results/income_deprivaiton.png', bbox_inches='tight')
```
