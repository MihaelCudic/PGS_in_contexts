```python
"""
    Code to analyze OR sensitivity to skewed PGS distribution
    Supplemental Figure S2
"""

import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
import statsmodels.api as sm
from scipy.special import expit
from scipy.stats import norm

import util.util as util
import util.pre_process as pre
import util.bootstrap_tools as btool
from scipy.stats import skewnorm

import matplotlib.pyplot as plt
import time

n_runs = 100

n_samps_ls = [40000, 100000, 350000]
prev = .05
n_covars = 16
top_pgs = .03
pgs_thresh = norm.ppf(1-top_pgs)
odds_ratio = 2.0

skew_ls = [-4, -2, 0, 2, 4]

n_samps_strata_ls = [1000, 2500, 5000, 10000, 25000, 50000]
```

# Run simulations

```python
import torch

def skew_normal(alpha, shape, device=None):
    delta = alpha / torch.sqrt(1 + alpha**2)
    z0 = torch.randn(shape, device=device)
    z1 = torch.randn(shape, device=device)
    return delta * torch.abs(z0) + torch.sqrt(1 - delta**2) * z1

def find_intercept(target_prev, beta_pgs, pgs):
    lo = torch.tensor(-5.0)
    hi = torch.tensor(0)
    for _ in range(60):
        mid = (lo + hi) * 0.5
        p = torch.special.expit(mid + beta_pgs * pgs).mean()
        hi = torch.where(p > target_prev, mid, hi)
        lo = torch.where(p <= target_prev, mid, lo)
    return (lo + hi) * 0.5

print('Find Beta and Intercept')
beta_pgs = []
intercept = []
for skew in skew_ls:
    pgs = skew_normal(torch.tensor(skew, dtype=torch.float32), (5000000,))
    pgs = (pgs - pgs.mean()) / pgs.std()
    
    thresh = torch.quantile(pgs, 1 - top_pgs)
    
    mu_hi = pgs[pgs > thresh].mean()
    mu_lo = pgs[pgs <= thresh].mean()
    beta_pgs_ = torch.log(torch.tensor(odds_ratio)) / (mu_hi - mu_lo)
    
    beta_pgs.append(beta_pgs_.item())
    
    intercept_ = find_intercept(prev, beta_pgs_, pgs)
    intercept.append(intercept_.item())
```

```python
import time
N_JOBS = 32

print('Run Simulations')
pain = np.full((len(n_samps_ls), len(skew_ls), len(n_samps_strata_ls), n_runs), np.nan)
subset = np.full_like(pain, np.nan)
inter = np.full_like(pain, np.nan)
inter_inter = np.full_like(pain, np.nan)
for i_samps, n_samps in enumerate(n_samps_ls):
    for i_skew, skew in enumerate(skew_ls):
        
        beta_pgs_ = beta_pgs[i_skew]
        intercept_ = intercept[i_skew]
        
        pgs = skew_normal(torch.tensor(skew, dtype=torch.float32), (n_runs, n_samps))
        pgs = (pgs-pgs.mean(1, keepdim=True))/pgs.std(1, keepdim=True)

        covars = torch.randn(n_runs, n_samps, n_covars+3)
        covars[:,:,:3] = 0

        prob = torch.special.expit(intercept_ + beta_pgs_ * pgs)
        phen = (torch.rand(n_runs, n_samps) < prob).float()

        phen_res = btool.efficient_linear_regression(covars[:,:,3:].numpy(), phen.numpy())['residuals']
        pgs_res = btool.efficient_linear_regression(covars[:,:,3:].numpy(), pgs.numpy())['residuals']

        t0 = time.time()
        for i_samps_strata, n_samps_strata in enumerate(n_samps_strata_ls):
            if (n_samps_strata >= n_samps):
                continue
            
            covars[:,:,3:] = 0
            
            pgs_ = pgs[:,:n_samps_strata]
            phen_ = phen[:,:n_samps_strata]
            prev_ = phen_.mean(1)

            phen_res_ = phen_res[:,:n_samps_strata]
            pgs_res_ = pgs_res[:,:n_samps_strata]

            pgs_thresh = torch.quantile(pgs_, 1 - top_pgs, dim=1, keepdim=True)
            high_pgs = (pgs > pgs_thresh)

            covars[:,:,2] = high_pgs
            covars[:,:n_samps_strata,1] = high_pgs[:, :n_samps_strata]
            covars[:,:n_samps_strata, 0] = 1
            
            # Pain et al Method
            r2_obs = btool.efficient_r2(pgs_res_, phen_res_)['r2']
            pain_ = btool.effecient_r2_to_risk(r2_obs, prev_.numpy(), top_pgs, None)['odds_ratio']
            pain[i_samps, i_skew, i_samps_strata, :] = pain_

            # Subset Method
            subset_params = btool.efficient_logistic_regression(covars[:,:n_samps_strata,2:], phen_, fetch_params=['weights'], n_jobs=N_JOBS)['weights']
            subset_or = np.exp(subset_params[:,0])
            subset[i_samps, i_skew, i_samps_strata, :] = subset_or

            # Interaction Method
            inter_params = btool.efficient_logistic_regression(covars[:,:,1:], phen, fetch_params=['weights'], n_jobs=N_JOBS)['weights']
            inter_or = np.exp(inter_params[:,0] + inter_params[:,1])
            inter[i_samps, i_skew, i_samps_strata, :] = inter_or

            # Interaction^2 Method
            inter_inter_params = btool.efficient_logistic_regression(covars, phen, fetch_params=['weights'], n_jobs=N_JOBS)['weights']
            inter_inter_or = np.exp(inter_inter_params[:,1] + inter_inter_params[:,2])
            inter_inter[i_samps, i_skew, i_samps_strata, :] = inter_inter_or
            
            print(f"# of Samps: {n_samps}, Skew: {skew}, Strata: {n_samps_strata}, TIME: {(time.time() - t0)/60}")
    
    np.save("results/pain.npy", pain)
    np.save("results/subset.npy", subset)
    np.save("results/inter.npy", inter)
    np.save("results/inter_inter.npy", inter_inter)
```

```python
from scipy.stats import gaussian_kde, norm
import seaborn as sns

fig, axs = plt.subplots(len(skew_ls), 1+len(n_samps_ls), figsize=(18, 2.5*len(skew_ls)))

# track y-limits for all non-first-column plots
ymins = []
ymaxs = []

for i_skew, skew in enumerate(skew_ls):
    x = np.linspace(-4, 4, 500)
    pgs = skew_normal(torch.tensor(skew, dtype=torch.float32), (100000,))
    pgs = (pgs - pgs.mean()) / pgs.std()

    axs[i_skew, 0].plot(x, norm.pdf(x), color='red', linewidth=2, label='Gauss.\n(α=0)')
    sns.kdeplot(pgs.numpy(), ax=axs[i_skew, 0], fill=True, alpha=0.4, label=f"α={skew}")
    axs[i_skew, 0].set_ylabel(axs[i_skew, 0].get_ylabel(), fontsize=18)
    axs[i_skew, 0].legend(fontsize=12, loc='upper right')
    axs[i_skew, 0].set_xlim(-5, 5)

pain_label = ["Pain et al. 2022", "strata", "high_pgs*inters.", "high_pgs*inters.+inters."]
for i_samps, n_samps in enumerate(n_samps_ls):
    for i_skew, skew in enumerate(skew_ls):
        pain_diff = abs(pain[i_samps, i_skew]-odds_ratio)
        strata_diff = abs(subset[i_samps, i_skew]-odds_ratio)
        inter_diff = abs(inter[i_samps, i_skew]-odds_ratio)
        inter_inter_diff = abs(inter_inter[i_samps, i_skew]-odds_ratio)
        
        means = [pain_diff.mean(1), strata_diff.mean(1), inter_diff.mean(1), inter_inter_diff.mean(1)]
        cis = [
            1.96 * abs(pain[i_samps, i_skew]-odds_ratio).std(1) / np.sqrt(n_runs),
            1.96 * abs(subset[i_samps, i_skew]-odds_ratio).std(1) / np.sqrt(n_runs),
            1.96 * abs(inter[i_samps, i_skew]-odds_ratio).std(1) / np.sqrt(n_runs),
            1.96 * abs(inter_inter[i_samps, i_skew]-odds_ratio).std(1) / np.sqrt(n_runs)
        ]
        x = np.arange(len(n_samps_strata_ls))
        width = 0.2
        axs[i_skew, i_samps+1].bar(x - 1.5*width, means[0], width, yerr=cis[0], label=pain_label[0])
        axs[i_skew, i_samps+1].bar(x - 0.5*width, means[1], width, yerr=cis[1], label=pain_label[1])
        axs[i_skew, i_samps+1].bar(x + 0.5*width, means[2], width, yerr=cis[2], label=pain_label[2])
        axs[i_skew, i_samps+1].bar(x + 1.5*width, means[3], width, yerr=cis[3], label=pain_label[3])
        axs[i_skew, i_samps+1].set_xticks(x)
        axs[i_skew, i_samps+1].set_xticklabels([str(v) for v in n_samps_strata_ls])
        #axs[i_skew, i_samps+1].legend(fontsize=11, loc='upper right')

        ymin, ymax = axs[i_skew, 1].get_ylim()
        ymins.append(ymin)
        ymaxs.append(ymax)

# apply shared y-axis to all columns except first
shared_min = min(ymins)
shared_max = max(ymaxs)

axs[-1,0].set_xlabel("Simulated PGS", fontsize=18)

for i in range(1, len(skew_ls)):
    axs[i,0].sharex(axs[0,0])

for j in range(1, 1+len(n_samps_ls)):
    axs[-1,j].set_xlabel("# of Samples in Strata", fontsize=18)
    for i in range(len(skew_ls)):
        axs[i,j].tick_params(axis='x', labelrotation=45, labelsize=14)
        for t in axs[i,j].get_xticklabels():
            t.set_ha("right")
        axs[i,j].tick_params(axis='y', labelsize=16)
        axs[i,j].set_ylabel("|GT OR - Pred. OR|", fontsize=16)
        

for i in range(len(skew_ls)):
    axs[i,0].tick_params(axis='both', labelsize=16)

for i in range(len(skew_ls)):
    for j in range(1, 1+len(n_samps_ls)):
        axs[i, j].set_ylim(shared_min, shared_max)
        axs[i, j].grid(axis='y', linestyle='--', alpha=0.6)

col_titles = ["", "Cohort Size: 40000", "Pop. Size: 100000", "Pop. Size: 350000"]
for j, title in enumerate(col_titles):
    axs[0, j].set_title(title, fontsize=20)

from matplotlib.lines import Line2D
handles = [
    Line2D([0], [0], color='C0', lw=8),
    Line2D([0], [0], color='C1', lw=8),
    Line2D([0], [0], color='C2', lw=8),
    Line2D([0], [0], color='C3', lw=8),
]

fig.legend(
    handles=handles,
    labels=pain_label,
    loc="lower center",
    bbox_to_anchor=(.65, -0.03 ),
    ncol=4,
    fontsize=14
)

fig.suptitle("OR Abs. Error with Different PGS Distribution Shapes", fontsize=24, fontweight='bold')

plt.tight_layout()
plt.savefig("nongauss.png", dpi=150, bbox_inches="tight")
plt.show()
```
